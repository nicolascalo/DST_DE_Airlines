```python
import pandas as pd
import folium
import os
import seaborn as sns
```

```python

path_call_parameter_file_folder = "call_parameter_lists"
call_parameter_csv_list = os.listdir(path_call_parameter_file_folder)


df_all = pd.DataFrame()

for call_parameter_csv in call_parameter_csv_list :
    
    print(call_parameter_csv)

    ### Import query parameters
    
    df_call_parameters = pd.read_csv(path_call_parameter_file_folder +"/"+call_parameter_csv).fillna('')
    
    df_all = pd.concat([df_all,df_call_parameters],ignore_index=True)


df_all = df_all[df_all['totalFlights'] != '']

df_date_range = df_all[(df_all['endRange'] == '2025-10-14T23:59:59Z')&(df_all['startRange'] == '2025-05-15T09:00:00Z')]
df_folium= df_date_range[['origin','destination','totalFlights']]
   
```

```python
df_folium = df_folium[df_folium['origin'] != df_folium['destination']]

df_folium_airport_stats_or = df_folium[['origin','totalFlights']]
df_folium_airport_stats_or.columns = ['iata', 'totalFlights']


df_folium_airport_stats_des = df_folium[['destination','totalFlights']]
df_folium_airport_stats_des.columns = ['iata', 'totalFlights']


df_folium_airport_all = pd.concat([df_folium_airport_stats_des,df_folium_airport_stats_or])

df_folium_airport_all.info()
```

```python


df_airports_desc = pd.read_csv("../df_iata_icao_wiki_final_world.csv").fillna('')
df_airports_desc.info()
```
